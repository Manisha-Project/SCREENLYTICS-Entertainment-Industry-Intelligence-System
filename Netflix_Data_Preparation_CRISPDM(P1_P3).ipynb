### PHASE 3: DATA QUALITY ASSESSMENT AND DATA PREPARATION

In this phase, we will systematically assess the Netflix dataset to identify and resolve data quality issues that may affect the accuracy, consistency, and reliability of further analysis.

The assessment and cleaning process will cover missing values, duplicate records, data types, inconsistent categorical values, invalid entries, formatting issues, and logical inconsistencies.

#### Objectives
1. **Identify**: Detect anomalies, missing entries, and structural errors (such as shifted columns).
2. **Investigate**: Determine the root cause of each data quality issue before modifying records.
3. **Prepare & Clean**: Apply targeted fixes, standardize formats, and handle missing values directly within this phase.

> **Note:** To preserve data integrity, the raw dataset (`df`) will remain untouched. All inspection and cleaning operations in this phase will be performed on a separate working copy (`df_clean`).

```python
# ============================================================
# PHASE 3: DATA QUALITY ASSESSMENT
# Project: Netflix Content Intelligence
# 1. MISSING VALUES ASSESSMENT
# Purpose: Identify missing values in each column.
# ============================================================

print("\n1. MISSING VALUES ASSESSMENT")

missing = df.isnull().sum()

missing_report = pd.DataFrame({
    'Missing Values': missing,
    'Missing Percentage': (missing / len(df) * 100).round(2)
})

display(missing_report.sort_values(
    by='Missing Values', ascending=False
))
```

```python
# Handle missing values in df_clean based on impact and field type

# 1. Director: Impute missing values with 'Unknown'
df_clean['director'] = df_clean['director'].fillna('Unknown')

# 2. Country: Impute missing values with 'Unknown' or mode (using 'Unknown' for categorical text fidelity)
df_clean['country'] = df_clean['country'].fillna('Unknown')

# 3. Cast: Impute missing values with 'Unknown'
df_clean['cast'] = df_clean['cast'].fillna('Unknown')

# 4. Rating & Date Added: Drop rows where critical metadata like rating or date_added are missing (very small percentage)
df_clean = df_clean.dropna(subset=['rating', 'date_added']).reset_index(drop=True)

# 5. Duration: Drop any remaining rows with missing duration if any survived previous cleaning steps
df_clean = df_clean.dropna(subset=['duration']).reset_index(drop=True)

print("Missing values successfully handled!")
print("Remaining missing values count:\n", df_clean.isnull().sum())
print(f"Current total rows: {len(df_clean)}")
```

```python
# ============================================================
# 2. DUPLICATE RECORDS ASSESSMENT
# Purpose: Identify completely duplicated records.
# ============================================================

print("\n2. DUPLICATE RECORDS ASSESSMENT")

print("Total Duplicate Rows:", df.duplicated().sum())

# print("\nDuplicate Records:")
# display(df[df.duplicated(keep=False)].head(10))
```

```python
# ============================================================
# 3. DATA TYPE VALIDATION
# Purpose: Identify columns with potentially incorrect data types.
# ============================================================

print("\n3. DATA TYPE VALIDATION")

dtype_report = pd.DataFrame({
    'Column': df.columns,
    'Data Type': df.dtypes.astype(str).values,
    'Non-Null Count': df.notnull().sum().values
})

display(dtype_report)
```

```python
# ============================================================
# 4. EMPTY AND BLANK STRING ASSESSMENT
# Purpose: Identify empty strings and whitespace-only values.
# ============================================================

print("\n4. EMPTY AND BLANK STRING ASSESSMENT")

blank_report = {}

for col in df.select_dtypes(include='object').columns:
    blank_report[col] = df[col].apply(
        lambda x: isinstance(x, str) and x.strip() == ''
    ).sum()

display(pd.Series(blank_report, name='Blank Values').to_frame())
```

```python
# ============================================================
# 5. CATEGORICAL CONSISTENCY ASSESSMENT
# Purpose: Examine distinct categorical values for inconsistencies.
# ============================================================

print("\n5. CATEGORICAL CONSISTENCY ASSESSMENT")

categorical_columns = [
    'type', 'rating', 'country', 'listed_in'
]

for col in categorical_columns:
    print(f"\nColumn: {col}")
    print("Unique Values:", df[col].nunique())
    print(df[col].value_counts(dropna=False).head(20))
```

```python
# ============================================================
# 6. WHITESPACE ASSESSMENT
# Purpose: Identify leading and trailing spaces in text columns.
# ============================================================

print("\n6. WHITESPACE ASSESSMENT")

for col in df.select_dtypes(include='object').columns:
    whitespace_count = df[col].apply(
        lambda x: isinstance(x, str) and x != x.strip()
    ).sum()

    print(f"{col}: {whitespace_count} records")
```

```python
# Filter and display rows where the title has extra spaces
whitespace_titles = df[df['title'].notnull() & (df['title'] != df['title'].str.strip())]
display(whitespace_titles)
```

```python
# ============================================================
# 7. DATE FORMAT VALIDATION
# Purpose: Identify invalid or unparseable date entries.
# ============================================================

print("\n7. DATE FORMAT VALIDATION")

date_values = pd.to_datetime(
    df['date_added'], errors='coerce'
)

invalid_dates = df['date_added'].notna() & date_values.isna()

print("Invalid Date Entries:", invalid_dates.sum())

display(df.loc[invalid_dates, ['date_added']].head(10))
```

```python
# 1. Clean the whitespace from the column
df['date_added'] = df['date_added'].str.strip()

# 2. Now convert them to proper datetime objects safely
df['date_added'] = pd.to_datetime(df['date_added'], errors='coerce')

# 3. Verify if any invalid dates remain
print("Remaining invalid dates:", df['date_added'].isna().sum())
```

```python
# Since df['date_added'] was converted to datetime in the previous cell,
# we temporarily cast it back to string to safely identify strings that failed to parse.
# Or we can simply filter the converted column for NaT values where it wasn't originally missing.

# Let's find rows where the parsed date is NaT (Not a Time) but wasn't originally null.
# We can also cast to string to perform any string-based inspection if needed:
remaining_bad_dates = df[df['date_added'].isna()]

# Display the problematic values (currently represented as NaT after conversion)
display(remaining_bad_dates[['date_added']])
```

```python
# Check the percentage of missing values in 'date_added'
missing_percentage = (df['date_added'].isna().sum() / len(df)) * 100
print(f"Missing percentage: {missing_percentage:.2f}%")
```

Date and time values are retained despite the low missing percentage, as they are important for analysing content addition trends and release patterns.

```python
# ============================================================
# 8 COMPREHENSIVE LOGICAL CONSISTENCY ASSESSMENT
# Purpose: Check duration formats, missing values, temporal logic,
# and field leakage directly on df
# ============================================================

print("\n=== COMPREHENSIVE LOGICAL CONSISTENCY ASSESSMENT ===")

# 1. Duration vs Type Inconsistencies
invalid_movies = df[
    (df['type'] == 'Movie') &
    (~df['duration'].astype(str).str.contains('min', na=False))
]

invalid_shows = df[
    (df['type'] == 'TV Show') &
    (~df['duration'].astype(str).str.contains('Season', na=False))
]

# 2. Missing Duration Check
missing_durations = df[df['duration'].isnull()]

# 3. Temporal Consistency: Date Added Year < Release Year
# Convert date_added to datetime to extract the calendar year
df['date_added_year'] = pd.to_datetime(
    df['date_added'].str.strip(), errors='coerce'
).dt.year

temporal_inconsistencies = df[
    df['date_added_year'] < df['release_year']
]

# 4. Field Leakage Check: Duration text stuck in the Rating column
rating_leakage = df[
    df['rating'].astype(str).str.contains('min', na=False)
]

# Print Summary Counts
print(f"1. Movies with invalid duration format: {len(invalid_movies)}")
print(f"2. TV Shows with invalid duration format: {len(invalid_shows)}")
print(f"3. Records with missing duration (NaN): {len(missing_durations)}")
print(f"4. Temporal violations (Date Added before Release Year): {len(temporal_inconsistencies)}")
print(f"5. Field leakage (Duration text stuck in Rating): {len(rating_leakage)}")

# Combine and display a sample if any issues are found
all_issues = pd.concat([invalid_movies, invalid_shows, missing_durations, temporal_inconsistencies, rating_leakage]).drop_duplicates()

if len(all_issues) > 0:
    print("\nSample Inconsistent / Problematic Records:")
    display(all_issues[['show_id', 'type', 'title', 'release_year', 'rating', 'duration']].head(10))
else:
    print("\nSuccess! All logical consistency checks passed with zero errors.")
```
